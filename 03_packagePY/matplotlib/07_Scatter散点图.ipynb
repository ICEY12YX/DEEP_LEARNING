```python
import matplotlib.pyplot as plt
import numpy as np

n = 1024
x = np.random.normal(0, 1, n)
y = np.random.normal(0, 1, n)
# 意思是生成1024个服从正态分布的随机点(均值为0，标准差为1)
t = np.arctan2(y, x)  # 用上面的x和y生成的每个点的角度array，将用于颜色映射

plt.scatter(x, y, s=75, c=t, alpha=0.5, cmap="rainbow")
# c=t意思是每个点的颜色根据其角度t来映射
# 疑问: 原理是啥，为什么颜色会根据角度变化
# 原理是matplotlib会根据c参数的值自动选择一个颜色映射表（colormap），
# 然后将每个点的c值映射到对应的颜色上。由于t是角度，所以不同角度的点会有不同的颜色。
# cmap参数指定颜色映射表，颜色映射表决定了不同的c值对应的颜色
plt.xlim(-1.5, 1.5)
plt.ylim(-1.5, 1.5)

# 隐藏坐标轴刻度(这里括号里一定加括号)
plt.xticks(())
plt.yticks(())
plt.colorbar()
# 显示颜色条

plt.show()
```

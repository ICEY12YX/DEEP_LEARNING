```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数

plt.figure()
plt.plot(x, y1)
# plt.scatter(x, y1) #可以试试看
# scatter相当于画线的时候换了个样式

ax = plt.gca()
ax.spines["right"].set_color("none")  # 隐藏右边的脊梁
ax.spines["top"].set_color("none")  # 隐藏上边的脊梁
# ax.xaxis.set_ticks_position("bottom")  # 将x轴的刻度设置在底部
# ax.yaxis.set_ticks_position("left")  # 将y轴的刻度设置在左边
# 上面两句其实不用写默认就是这么做的
ax.spines["bottom"].set_position(("data", 0))  # 将x轴移动到y=0的位置
ax.spines["left"].set_position(("data", 0))  # 将y轴移动到x=0的位置

# 添加标注
x0 = 1
y0 = 2 * x0 + 1
plt.scatter(x0, y0, s=50, color="r")  # 标注点
# 因为x0就一个点,输出y0也就一个点  s是散点的大小size
plt.plot([x0, x0], [y0, 0], "k--", lw=2.5)
# 意思是画一条从标注点到x轴的虚线，'k--'表示黑色虚线，lw=2.5表示线宽为2.5
# [x0,x0],[y0,0] 可以理解成从(x0,y0)到(x0,0)画一条竖直的虚线
# 疑问:这啥原理

# 方一:
plt.annotate(
    f"(2x+1={y0})",
    xy=(x0, y0),
    xycoords="data",
    xytext=(+30, -30),
    textcoords="offset points",
    fontsize=16,
    arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.2"),
)  # 添加注释
# 每个参数意思是
# f"(2x+1={y0})" : 注释文本
# xy=(x0, y0) : 注释指向的点
# xycoords="data" : xy的坐标系为数据坐标
# xytext=(+30, -30) : 注释文本的位置，相对于xy的偏移
# textcoords="offset points" : xytext的坐标系为偏移点
# fontsize=16 : 注释文本的字体大小
# arrowprops=dict(arrowstyle="->", connectionstyle="arc3,rad=.2") : 注释箭头的属性

# 方二: 使用文本注释,就是加个文本框
plt.text(-3.7, 3, f"This is some text", fontsize=16, color="g")
plt.text(-3.7, 6, f"This is some text", fontdict={"fontsize": 16, "color": "b"})
# 每个参数意思是
# -3.7,3 : 文本的位置
# f"This is some text" : 文本内容
# fontsize=16 : 文本字体大小
# color="g" : 文本颜色为绿色

plt.show()
```

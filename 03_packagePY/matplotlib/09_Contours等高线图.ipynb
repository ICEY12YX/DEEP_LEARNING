```python
import matplotlib.pyplot as plt
import numpy as np


def f(x, y):
    # the height function
    return (1 - x / 2 + x**5 + y**3) * np.exp(-(x**2) - y**2)


# 该函数表示一个高度函数，用于生成等高线图
# exp意思是指数函数，即e的幂次方

# 登高线图的数据其实是在Z轴上的,底图还有个X-Y平面(我们称为网格,这是我们下面生成的)
# 网格就是等距生成就好(学一下下面的思路)
n = 256
x = np.linspace(-3, 3, n)
y = np.linspace(-3, 3, n)
# 生成网格点坐标矩阵
X, Y = np.meshgrid(x, y)

# 使用 plt.contourf 填充等高线
# 从这步开始有图像了
plt.contourf(X, Y, f(X, Y), 8, alpha=0.75, cmap=plt.cm.hot)
# plt.contourf(X, Y, f(X, Y), 8, alpha=0.75, cmap="cool")
# 意思是绘制填充的等高线图，8表示等高线的数量，alpha表示透明度，cmap表示颜色映射方案
# plt.cm.hot 或者写成"hot"
# X, Y, f(X, Y)意思是网格上x,y位置上的高度值

# 使用 plt.contour 添加等高线,那条线
C = plt.contour(X, Y, f(X, Y), 8, colors="black", linewidths=0.5)
# C = plt.contour(X, Y, f(X, Y), 30, colors="black", linewidths=0.5)
# 8的意思是等高线的数量
# 0是分2部分,那8就是分10部分

# 添加标签
plt.clabel(C, inline=True, fontsize=10)
# 每个参数意思是
# C: 等高线对象
# inline=True: 标签是否嵌入在等高线内
# fontsize=10: 标签字体大小

plt.show()
```

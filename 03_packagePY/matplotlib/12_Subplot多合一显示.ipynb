```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()

plt.subplot(2, 2, 1)
# 创建一个2行2列的子图网格(就是总共有4个子图位置)，并激活第一个子图(就是下面的线展示在这个子图中)
plt.plot([0, 1], [0, 1])  # 在第一个子图中绘制一条直线
# 这个用法在05有解释

plt.subplot(2, 2, 2)  # 激活第二个子图
plt.plot([0, 1], [0, 2])  # 在第二个子图中绘制一条直线

plt.subplot(223)  # 也可以偷懒写成这样
plt.plot([0, 1], [0, 3])

plt.subplot(224)
plt.plot([0, 1], [0, 4])

plt.show()  # 显示所有子图
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure()

plt.subplot(2, 1, 1)
plt.plot([0, 1], [0, 1])

plt.subplot(2, 3, 4)
plt.plot([0, 1], [0, 2])

plt.subplot(235)
plt.plot([0, 1], [0, 3])

plt.subplot(236)
plt.plot([0, 1], [0, 4])
# 为什么最后一个数字填4,5,6 用下面的代码对比就知道了
# 大概就是 原意 第一行分1格 第二行分3格
# 但是程序还是按照,2x3(最大的分割数)分,所以要填4,5,6

plt.show()  # 显示所有子图
```

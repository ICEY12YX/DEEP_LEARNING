```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 0.1 * x  # 线性函数

plt.figure()
plt.plot(x, y1, linewidth=10, zorder=1)  # 当线太粗的时候挡住了xy轴的数据
plt.ylim(-2, 2)

ax = plt.gca()
ax.spines["right"].set_color("none")  # 隐藏右边的脊梁
ax.spines["top"].set_color("none")  # 隐藏上边的脊梁
# ax.xaxis.set_ticks_position("bottom")  # 将x轴的刻度设置在底部
# ax.yaxis.set_ticks_position("left")  # 将y轴的刻度设置在左边
# 上面两句其实不用写默认就是这么做的
ax.spines["bottom"].set_position(("data", 0))  # 将x轴移动到y=0的位置
ax.spines["left"].set_position(("data", 0))  # 将y轴移动到x=0的位置

# 线遮住轴上的刻度标签,只能笨办法
# 拿到所有的刻度标签并设置其字体大小和颜色
for label in ax.get_xticklabels() + ax.get_yticklabels():
    label.set_fontsize(12)
    label.set_bbox(dict(facecolor="white", edgecolor="none", alpha=0.7, pad=10))
    label.set_color("r")
    label.set_zorder(99)
# bbox就是给刻度标签设置背景框的属性
# 每个参数意思是
# facecolor: 背景颜色
# edgecolor: 边框颜色
# alpha: 透明度
# pad: 内边距
# 仅通过设置alpha不能达到label前置（仍被覆盖）
# 需要同时设置zorder
plt.show()
```

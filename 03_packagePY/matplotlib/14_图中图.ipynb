```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure()
x = [1, 2, 3, 4, 5, 6, 7]
y = [1, 3, 4, 2, 5, 8, 6]

# 创建主图
# left, bottom, width, height = 0.1, 0.1, 0.1, 0.8 #试一下这个参数
# left, bottom, width, height = 0.1, 0.8, 0.8, 0.8 #试一下这个参数
left, bottom, width, height = 0.1, 0.1, 0.8, 0.8
ax1 = fig.add_axes([left, bottom, width, height])
# 意思是创建一个新的Axes对象,位置由[left, bottom, width, height]指定,相对于整个图的坐标系,范围是0到1
# 疑问: 范围0到1是默认的吗?
# 答:是的,可以理解为百分比,0.8就是占整个窗口的80%大小(长宽)
# left, bottom 0.1 表示主图距离整个窗口左边和下边各占10%的位置(改这个,图中图的位置也会相应变化,疑问:为什么???)
ax1.plot(x, y, "r")  # r是加上红色
ax1.set_xlabel("X")
ax1.set_ylabel("Y")
ax1.set_title("Main Plot")
# 还是那个疑问,之前简单画图的时候都没有用到Axes对象,直接用plt.plot就可以了,这是因为plt.plot会自动获取当前的Axes对象,而这里我们显式创建了一个Axes对象,所以需要通过ax1来调用绘图方法。 ???
# 好像百分比都是按照整个图来算的,那为什么主图left, bottom改变,图中图位置会变???

# 创建图中图的Axes对象
left, bottom, width, height = 0.2, 0.6, 0.25, 0.25
ax2 = fig.add_axes([left, bottom, width, height])
ax2.plot(x, y, "b")  # b是蓝色

plt.axes([0.6, 0.2, 0.25, 0.25])
plt.plot(x[::-1], y, "g")  # g是绿色
# [::-1]意思是将列表反转,原理是使用切片语法,从头到尾以步长-1取元素
# 疑问: 还是没太懂

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 方1: subplot2grid
plt.figure()
ax1 = plt.subplot2grid((3, 3), (0, 0), colspan=3)
# 意思是 创建子图网格为3行3列,当前子图在0行0列,并且跨越3列
ax1.plot([0, 1], [0, 1])  # 在ax1子图上绘制数据
ax1.set_title("Ax1")  # 给ax1子图设置标题
ax1.set_ylabel("Y轴")  # 给ax1子图设置Y轴标签
# 之前的写法是plt.ylabel("Y轴")  那这里函数名要加set_, 同时注意是新创建的网格ax1在调用

ax2 = plt.subplot2grid((3, 3), (1, 0), colspan=2)
# 注意每个子图都是一个Axes对象,画一个新的子图时要创建新的Axes对象
# 疑问:啥是Axes对象?
ax3 = plt.subplot2grid((3, 3), (1, 2), rowspan=2)
ax4 = plt.subplot2grid((3, 3), (2, 0))  # 默认占一小格
ax5 = plt.subplot2grid((3, 3), (2, 1))

plt.show()  # 显示所有子图
```

```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 方2: gridspec (比上面写法简单)
plt.figure()
gs = gridspec.GridSpec(3, 3)
# 创建一个3行3列的网格布局对象
ax1 = plt.subplot(gs[0, :])
# 这个子图占据第一行的所有列
ax2 = plt.subplot(gs[1, :2])
# 这个子图占据第二行的前两列
ax3 = plt.subplot(gs[1:, 2])
# 这个子图占据第二行到后面的行, 并且在第三列
ax4 = plt.subplot(gs[2, 0])
# 这个子图占据第三行的第一列
ax5 = plt.subplot(gs[2, 1])

plt.show()  # 显示所有子图
```

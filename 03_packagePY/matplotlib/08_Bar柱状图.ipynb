```python
import matplotlib.pyplot as plt
import numpy as np

n = 12
x = np.arange(n)
y1 = (1 - x / float(n)) * np.random.uniform(0.5, 1.0, n)
y2 = (1 - x / float(n)) * np.random.uniform(0.5, 1.0, n)
# uniform意思是在指定范围内生成均匀分布的随机数
# np.random.uniform(low, high, size)分别表示下限、上限和生成的随机数个数
# 疑问:上面算式想表达什么
print(x)
print(y1)
print(y2)

plt.bar(x, +y1, facecolor="blue", edgecolor="black")
plt.bar(x, -y2, facecolor="red", edgecolor="black")
# +/-表示正负方向的柱状图

for x, y in zip(x, y1):
    plt.text(x, y + 0.05, "%.2f" % y, ha="center", va="bottom")
# 在x,y点偏上面一点设置text
# 其他参数意思是
# ha='center' 表示水平对齐方式为居中
# va='bottom' 表示垂直对齐方式为底部

# for x, y in zip(x, y2):
#     plt.text(x, -y - 0.05, "-%.2f" % y, ha="center", va="top")
# 疑问:报错???

plt.xlim(-0.5, n)
# plt.xticks(())
plt.ylim(-1.25, 1.25)
# plt.yticks(())

plt.show()
```

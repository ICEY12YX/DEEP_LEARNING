```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数
y2 = x**2  # 二次函数

plt.figure()
# 意思是创建一个新的图像窗口,figure理解为窗口
# 之前不写figure的话,默认创建一个窗口,所有的plot都会在同一个图像窗口里显示
plt.plot(x, y1)

plt.figure()
# 两句figure意味着我们创建了两个独立的图像窗口,每个窗口可以绘制不同的图像
plt.plot(x, y2)

plt.show()
# 两句figure只用一次show
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数
y2 = x**2  # 二次函数

plt.figure()
plt.plot(x, y1)

# 可以调节参数
plt.figure(num=99, figsize=(8, 6))
# 意思是创建一个编号为99的图像窗口,并设置窗口大小为8x6英寸
plt.plot(x, y1, color="r", linewidth=5, linestyle="--")
# 意思是绘制一条红色的虚线,线宽为2
plt.plot(x, y2)
# 两条线放一个figure里

plt.show()
```

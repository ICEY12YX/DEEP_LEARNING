```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10, 0.1)
y1 = 0.05 * x**2
y2 = -1 * y1
# print(x, y1, y2)

fig, ax1 = plt.subplots()
# subplots其实返回一个包含Figure对象和Axes对象的元组-> tuple[Figure, Axes] (悬停查看),这里我们用fig接收Figure对象,ax1接收Axes对象
ax1.plot(x, y1, "b--")  # b--意思是蓝色虚线
ax1.set_xlabel("X")
ax1.set_ylabel("Y1", color="b")

ax2 = ax1.twinx()
# 意思是创建一个与ax1共享x轴但独立y轴的Axes对象,这样可以在同一张图上绘制两个y轴不同的数据
ax2.plot(x, y2, "r-")
ax2.set_ylabel("Y2", color="r")
# 坐标轴上的数字是根据各自的y轴范围自动生成的
# y2看上去会有点奇怪,因为y2和y1坐标轴数值不一样,实际上放在一张图上是下面这样的

plt.figure()
plt.plot(x, y1, "b--")
plt.plot(x, y2, "r-")
plt.xlabel("X")
plt.ylabel("Y")
plt.ylim(-8, 8)
plt.title("Combined Plot")

plt.show()
```

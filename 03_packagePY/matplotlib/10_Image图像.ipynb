```python
import matplotlib.pyplot as plt
import numpy as np

# image data(当做图像的像素值)
a = np.array(
    [
        0.313660827978,
        0.365348418405,
        0.423733120134,
        0.365348418405,
        0.439599930621,
        0.525083754405,
        0.423733120134,
        0.525083754405,
        0.651536351379,
    ]
).reshape(3, 3)

# 显示图像
plt.imshow(a, interpolation="nearest", cmap="bone", origin="lower")
# plt.imshow(a, interpolation="bilinear", cmap="hsv", origin="lower")
# 每个参数意思是
# a: 图像数据(像素值矩阵)
# interpolation: 图像的插值方法,就是图像怎么处理两个像素之间如何计算中间值
# cmap: 颜色映射方案
# origin: 原点位置


# 隐藏刻度
plt.xticks(())
plt.yticks(())
# 添加颜色条
plt.colorbar(shrink=0.9)
# shrink参数控制颜色条的缩放比例(就是显示的大小)

plt.show()
```

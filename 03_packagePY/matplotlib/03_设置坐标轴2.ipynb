```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数
y2 = x**2  # 二次函数

plt.figure()
plt.plot(x, y1, color="r", linewidth=1, linestyle="--")
plt.plot(x, y2)

plt.xlim(-1, 2)  # 设置x轴的范围
plt.ylim(-2, 3)  # 设置y轴的范围

new_ticks = np.linspace(-1, 2, 5)  # 设置新的刻度
plt.xticks(new_ticks)  # 设置x轴的刻度

# gca 意思是 "get current axis", 用于获取当前的坐标轴对象，可以用来进一步设置坐标轴的属性
# 脊梁,也可以理解成边框
ax = plt.gca()
ax.spines["right"].set_color("none")  # 隐藏右边的脊梁
ax.spines["top"].set_color("none")  # 隐藏上边的脊梁
# ax.xaxis.set_ticks_position("top")  # 将x轴的刻度设置在顶部(四个边框,顶部那个位置)
# ax.yaxis.set_ticks_position("right")  # 将y轴的刻度设置在右边

ax.xaxis.set_ticks_position("bottom")  # 将x轴的刻度设置在底部
ax.yaxis.set_ticks_position("left")  # 将y轴的刻度设置在左边
# 上面两句其实不用写默认就是这么做的

ax.spines["bottom"].set_position(("data", 0))  # 将x轴移动到y=0的位置
ax.spines["left"].set_position(("data", 0))  # 将y轴移动到x=0的位置
# set_position 里,"data" 表示根据数据坐标来设置脊梁的位置,还能设置为"outward"表示向外偏移, "axes"表示根据轴的坐标来设置脊梁的位置
# 疑问: 上面的轴怎么知道移动到y或x轴的0的位置的?
# 应该这么理解,bottom轴(不管现在是x轴还是y轴), 移动到另一条轴的0位置, left轴同理, 移动到另一条轴的0位置

# 举例
# ax.spines["bottom"].set_position(("outward", 20))  # 将x轴向外偏移20个点
# ax.spines["bottom"].set_position(("outward", -20))  # 将x轴向内偏移20个点
# ax.spines["left"].set_position(("axes", 0.1))  # 将y轴移动到轴的0.1位置
# ax.spines["left"].set_position(("axes", -0.1))  # 将y轴移动到轴的-0.1位置
# 0.1表示将y轴移动到另一条轴的10%位置，-0.1表示将y轴移动到轴的-10%位置

plt.show()
```

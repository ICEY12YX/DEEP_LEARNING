```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数
y2 = x**2  # 二次函数

plt.figure()

plt.xlim(-1, 2)  # 设置x轴的范围
plt.ylim(-2, 3)  # 设置y轴的范围
new_ticks = np.linspace(-1, 2, 5)  # 设置新的刻度
plt.xticks(new_ticks)  # 设置x轴的刻度

# 也可以先设置xy轴相关,再搞条线
# 设置图例
plt.plot(x, y1, color="r", linewidth=1, linestyle="--", label="yuan")
plt.plot(x, y2, label="fan")

plt.legend()  # 显示图例 (不写是不显示上面设置的label的)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2 * x + 1  # 线性函数
y2 = x**2  # 二次函数

plt.figure()

plt.xlim(-1, 2)  # 设置x轴的范围
plt.ylim(-2, 3)  # 设置y轴的范围
new_ticks = np.linspace(-1, 2, 5)  # 设置新的刻度
plt.xticks(new_ticks)  # 设置x轴的刻度

# 设置图例
(l1,) = plt.plot(x, y1, color="r", linewidth=1, linestyle="--", label="yuan")
(l2,) = plt.plot(x, y2, label="fan")
# 要传进legend里,上面l1和l2后面要加, 括号随便加不加
# 显示图例
plt.legend(handles=[l1, l2], loc="best")
# loc表示图例的位置, "best"表示自动选择最佳位置
# plt.legend(handles=[l1],labels=["222", "111"], loc="best")
# 也可以直接使用label参数, 会覆盖上面自己设置的label
# 可以当做这里的label就是给上面两条线增加属性
# 一条线的图例,但是设置了两个label,可能报错,最好不要
plt.show()
```

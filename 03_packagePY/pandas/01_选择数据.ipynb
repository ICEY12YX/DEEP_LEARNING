```python
import pandas as pd
import numpy as np

dates = pd.date_range("20230101", periods=6)
df = pd.DataFrame(np.arange(24).reshape(6, 4), index=dates, columns=list("ABCD"))
# print(df)

# 取数据
print(df["A"])  # 取A列
print(df.A)  # 和上面一样的
print(df[0:3])  # 取前3行
print(df["20230101":"20230103"])  # 取前3行 ,这个时候就没有左闭右开了
# index不能像column一样使用,下面这两个会报错
# 可以记成表格里一般index就是数字,df.1 这样取也太怪了
# print(df["20230101"])
# print(df."20230101")
# print(df["20230101", "B"]) 这个也会报错
# 看下面的loc,所以说用index这列选择的时候,要搭配loc

# select by label: loc
print(df.loc["2023-01-01"])  # 取第一行
print(df.loc["20230101"])  # 取第一行 一样的
print(df.loc[:, "A"])  # 取A列
# loc的第一个参数是行的条件,第二个参数是列的条件
# 前面的:表示取所有行,后面的"A"表示取A列
# 疑问: 为什么取"20230101"不用加:冒号  可能因为loc是按标签取的,而这个标签就是指index这列的内容
print(df.loc["20230101", ["A", "B"]])  # 取第一行的A列和B列

# select by position: iloc (和np差不多了)
print(df.iloc[3])  # 取第4行
print(df.iloc[3, 1])  # 取第4行第2列
print(df.iloc[3:5, 0:2])  # 取第4行到第5行的第1列到第2列 (切片)
print(df.iloc[[1, 2, 4], [0, 2]])  # 取第2行、第3行、第5行的第1列和第3列

# mixed selection: ix (已经被弃用,不建议使用)
# print(df.ix[0:3, ["A", "C"]])  # 弃用了,用了会报错

# boolean indexing
print(df[df.A > 8])  # 取A列大于8的行
```

```python
import pandas as pd
import numpy as np

dates = pd.date_range("20230101", periods=6)
df = pd.DataFrame(np.arange(24).reshape(6, 4), index=dates, columns=list("ABCD"))
print(df)

# 其实知道怎么选择就知道怎么改了
df.iloc[2, 2] = 111
df.loc["20230101", "B"] = 222

print(df.A > 4)
print(type(df.A > 4))  # 输出是Series
df[df.A > 4] = 0
# 这个是因为df.A > 4的结果是一个布尔型的Series,得到的结果是每一行的
# 所以把这个给df,那就一整行一起处理了(变0)
print(df)
```

```python
import pandas as pd
import numpy as np

dates = pd.date_range("20230101", periods=6)
df = pd.DataFrame(np.arange(24).reshape(6, 4), index=dates, columns=list("ABCD"))
print(df)


# df.B[df.A > 4] = 0
# 这样就是只处理B列,其他列不变(因为只把这个bool型的Series给B列了,其他列不变)
# 疑问: 上面的改法好像会有警告,说是链式赋值,那有没有更好的方法呢?
df.loc[df.A > 4, "B"] = 0
# 这样改,意思是: 先选出A列大于4的行,然后在这些行里选出B列,然后把这些B列的值改为0
# loc的第一个参数是行的条件,第二个参数是列的条件
print(df)

# 新加一列
df["F"] = np.nan
print(df)

# df["E"] = pd.Series([1, 2, 3, 4, 5,6], index=pd.date_range("20230101", periods=6))
df["E"] = pd.Series([1, 2, 3, 4, 5, 6], index=df.index)
# 或者让这个Series的index直接和df的index对齐
# df["E"] = pd.Series([1, 2, 3, 4, 5], index=pd.date_range("20230101", periods=6))
# 这样写,没对齐,会报错
# df["E"] = pd.Series([1, 2, 3, 4, 5, 6])
# 疑问:上面那样写,E列会全为nan
print(df)
```

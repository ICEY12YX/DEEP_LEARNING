```python
import pandas as pd
import numpy as np

# 之前学的字典的方式
left = pd.DataFrame(
    {
        "key": ["K0", "K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)
right = pd.DataFrame(
    {
        "key": ["K0", "K1", "K2", "K3"],
        "A": ["A0", "A1", "A2", "A3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    }
)
print(left)
print(right)

result = pd.merge(left, right, how="inner", on="key")
print(result)
# 意思是按照key列进行内连接,也就是取两个表中key列有共同值的行,进行合并,如果有重复的列名,会自动加上后缀_x和_y来区分
# 默认的how是inner,也就是内连接,意思就是上面这个
# outer是外连接,意思就是取两个表中key列的所有值,如果某个表中没有对应的值,则用NaN填充
print(result)
```

```python
import pandas as pd
import numpy as np

# 之前学的字典的方式
left = pd.DataFrame(
    {
        "key": ["K0", "K0", "K1", "K2"],
        "key2": ["K0", "K1", "K0", "K1"],
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
    }
)
right = pd.DataFrame(
    {
        "key": ["K0", "K1", "K1", "K2"],
        "key2": ["K0", "K0", "K0", "K0"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    }
)

result = pd.merge(left, right, how="inner", on=["key", "key2"])
# inner只考虑相同的部分,两列考察的话,就是两列完全一样的
# 但是可以观察到上面K1,K0有两行没对齐,所以输出的时候就有两行,同时一行输出的时候,会去找k1,k0的行,所以不会输出nan
print(result)
result = pd.merge(left, right, how="outer", on=["key", "key2"])
print(result)
```

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({"col1": [0, 1], "col_left": ["a", "b"]})
df2 = pd.DataFrame({"col1": [1, 2, 2], "col_right": [2, 2, 2]})
print(df1)
print(df2)

result = pd.merge(df1, df2, on="col1", how="outer", indicator=True)
print(result)
# indicator参数可以显示每一行是来自哪个表的,左表为left_only,右表为right_only,两表都有为both
result = pd.merge(df1, df2, on="col1", how="outer", indicator="indicator_column")
print(result)
# 不写True,可以自定义列名
```

```python
import pandas as pd
import numpy as np

# 之前学的字典的方式
left = pd.DataFrame(
    {"A": ["A0", "A1", "A2", "A3"], "B": ["B0", "B1", "B2", "B3"]},
    index=["K0", "K1", "K2", "K3"],
)
right = pd.DataFrame(
    {"C": ["C0", "C1", "C2", "C3"], "D": ["D0", "D1", "D2", "D3"]},
    index=["K0", "K1", "K2", "K3"],
)
# 自定义index
print(left)
print(right)

result = pd.merge(left, right, left_index=True, right_index=True, how="outer")
print(result)
# left_index,right_index参数默认是False
# True表示使用index进行合并,False表示使用列名进行合并

# result = pd.merge(left, right,how="outer")
# print(result)
# 报错,因为没有相同的列名,所以无法进行合并
```

```python
import pandas as pd
import numpy as np

boys = pd.DataFrame({"k": ["K0", "K1", "K2"], "age": [1, 2, 3]})
girls = pd.DataFrame({"k": ["K0", "K0", "K3"], "age": [4, 5, 6]})

result = pd.merge(boys, girls, on="k", how="inner")
print(result)
# 发现,age列重复了,默认会加上后缀_x和_y,但是可以自定义后缀,就是用suffixes参数
result = pd.merge(boys, girls, on="k", how="inner", suffixes=["_boy", "_girl"])
print(result)
result = pd.merge(boys, girls, on="k", how="inner")
print(result)
```

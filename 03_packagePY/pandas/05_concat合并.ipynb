```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame(np.ones((3, 4)) * 0, columns=list("ABCD"))
df2 = pd.DataFrame(np.ones((3, 4)) * 1, columns=list("ABCD"))
df3 = pd.DataFrame(np.ones((3, 4)) * 2, columns=list("ABCD"))
# *0,*1的意思是创建一个3行4列的DataFrame，所有元素都为0或1或2，列名为"A"、"B"、"C"和"D"。
# index,和columns都选择省略
# print(df1)
# print(df2)
# print(df3)

# 合并
result = pd.concat([df1, df2, df3], axis=0)
# axis=0表示列方向合并,即将df1、df2、df3按行拼接在一起
print(result)
# 但是发现索引重复了,所以我们可以使用ignore_index=True来重新索引,意思是忽略原来的索引,重新生成一个新的索引
result = pd.concat([df1, df2, df3], axis=0, ignore_index=True)
print(result)
```

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame(np.ones((3, 4)) * 0, columns=list("ABCD"))
df2 = pd.DataFrame(np.ones((3, 4)) * 1, columns=list("BCDE"))
# print(df1)
# print(df2)

# join ['inner', 'outer'], 默认是 'outer'
# inner表示取交集,outer表示取并集,默认是outer
result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)
# 发现,列名不一样,所以会出现NaN

# 所以用到join因为outer取并集,会把所有的列名都保留下来,但是df1没有E列,所以E列的值为NaN
result = pd.concat([df1, df2], axis=0, join="outer", ignore_index=True)
print(result)
# inner取交集,A,E列就会被删掉
result = pd.concat([df1, df2], axis=0, join="inner", ignore_index=True)
print(result)
```

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame(np.ones((3, 4)) * 0, index=list("123"), columns=list("ABCD"))
df2 = pd.DataFrame(np.ones((3, 4)) * 1, index=list("234"), columns=list("BCDE"))
# print(df1)
# print(df2)

result = pd.concat([df1, df2], axis=1, ignore_index=True)
print(result)
# 这次是index不一样了,所以拼接还是会有nan

# 用到join_axes
# result = pd.concat([df1, df2], axis=1, join_axes=[df1.index])
# 意思是按照df1的index来拼接,所以df2的index为4的那一行会被舍弃,但是df2没有index为1的那一行,所以这一行的值为NaN
# 但是这个方法已经被弃用了,所以我们可以使用reindex来实现同样的效果
# 疑问: reindex前面那个点就这样加?
result = pd.concat([df1, df2.reindex(df1.index)], axis=1, ignore_index=True)
print(result)
```

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame(np.ones((3, 4)) * 0, index=list("123"), columns=list("ABCD"))
df2 = pd.DataFrame(np.ones((3, 4)) * 1, index=list("123"), columns=list("ABCD"))

# append()方法用于将一个DataFrame追加到另一个DataFrame的末尾,返回一个新的DataFrame,原来的DataFrame不变
# 但是append()方法已经被弃用了,所以我们可以使用concat()方法来实现同样的效果
# result = df1.append(df2, ignore_index=True)
# print(result)

s = pd.Series([1, 2, 3], index=list("123"))
# 怎么把series加到dataframe里,可以用concat,但是要注意axis=1,因为默认是axis=0,所以会报错
result = pd.concat([df1, s], axis=1)
# 列名是默认的
print(result)

s = pd.Series([1, 2, 3, 4], index=list("ABCD"))
s1 = pd.Series([1, 2, 3], index=list("123"))
# 如果要用series变为一行加到dataframe里,可以用to_frame()方法把series变为dataframe,然后再concat
result = pd.concat([df1, s.to_frame().T], axis=0, ignore_index=True)
result = pd.concat([result, s1.to_frame().T], axis=0, ignore_index=True)
print(result)
```

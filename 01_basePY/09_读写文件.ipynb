```python
text = "This is a sample text for testing the code.\nIt contains multiple lines.\nThis is the third line."

my_file = open("sample.txt", "w")  # 打开文件, 如果文件不存在, 会创建一个新文件
# open的第二个参数 "w" 表示写入模式, 如果文件已经存在, 会覆盖原有内容
# 疑问: 另外还有哪些
# 只写sample.txt应该是以现在的位置为基础的相对路径
my_file.write(text)  # 写入内容到文件
my_file.close()  # 关闭文件  必须关上!
# 疑问: 为什么必须关上?
# 好像不关上jypter也不会报错
```

```python
append_text = "\nThis is an appended line.(追加文字)"
my_file = open("sample.txt", "a")  # 打开文件, 如果文件不存在, 会创建一个新文件
#  "a" 表示追加模式, 如果文件已经存在, 会在原有内容后追加新内容
# 疑问: 是不是不管什么模式,打开文件文件不存在就创建?
my_file.write(append_text)  # 写入内容到文件
my_file.close()
```

```python
my_file = open("sample.txt", "r")  # 打开文件, 如果文件不存在, 会创建一个新文件
# content = my_file.read()  # 读取文件内容
# print(content)  # 打印文件内容

content_lines = my_file.readlines()  # 读取文件内容, 按行存储到列表中
# 疑问: 此时content_lines应该是空的相当于一个列表?
print(content_lines)
# 注意如果上面content的两句不注释掉,这里就读不出来东西,输出空数组
# 注释掉就能正常输出打印(我猜大概是指针的问题)
# 疑问:为什么会出现上面的问题?
# 输出类似['This is a sample text for testing the code.\n', 'It contains multiple lines.\n', 'This is the third line.\n', 'This is an appended line.(追加文字)']
# 你可以发现\n也被分到每行元素中(反正一行就是一个元素)

my_file.close()
```

# continue & break

```python
a = True
while a:
    b = input("type something")
    if b == "1":
        a = False
    else:
        pass  # 这啥意思
    print("still in while")
print("finish")
```

```python
while True:
    b = input("type something")
    if b == "1":
        break
    else:
        continue
    print("still in while")  # 注意pass和continue不一样(对比上下两端代码)
print("finish")
```

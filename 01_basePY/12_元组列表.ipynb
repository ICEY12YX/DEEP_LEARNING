```python
a_tuple = (1, 2, 3, 4, 5)
a_list = [77, 88, 99, 100]
# 元组, 元组是不可变的序列, 元组中的元素可以是任意类型, 元组中的元素可以重复
# 列表, 列表是可变的序列, 列表中的元素可以是任意类型, 列表中的元素可以重复
# 注意一个是(),一个是[]

# i是数组元素
for i in a_tuple:
    print(i)

for i in a_list:
    print(i)

# i是索引
# range(len(a_tuple))意思是, range()函数生成一个整数序列, len()函数返回序列的长度, 所以range(len(a_tuple))生成一个从0到len(a_tuple)-1的整数序列, 也就是a_tuple的索引序列
# range只有一个参数, 就是终止值, 生成的整数序列从0开始, 所以range(len(a_tuple))生成的整数序列就是a_tuple的索引序列
# 疑问: 上面那个对吗
for index in range(len(a_tuple)):
    print(a_tuple[index])

# 上面这个for,用a_list也是完全一样的效果
for index in range(len(a_list)):
    print(a_list[index])
```

```python
c_list = [1, 2, 3, 4, 5]
c_list.insert(2, 99)  # 在索引2的位置插入99
c_list.insert(2, 99)  # 在索引2的位置插入99
c_list.insert(2, 99)  # 在索引2的位置插入99
print(c_list)

c_list.remove(99)
# 删除第一个出现的99(会找到这个99, 然后删除它, 如果有多个99, 只会删除第一个)
print(c_list)  # 输出: [1, 2, 3, 4, 5]
print(c_list[-1])  # 输出: 5, -1表示最后一个元素, -2表示倒数第二个元素, 以此类推
print(c_list[0:3])  # 输出: [1, 2, 3], 输出0,1,2位元素,不包括3号位元素
print(c_list[3:])  # 3号位到最后
print(c_list[-2:])  # 倒数第二位到最后
print(c_list.index(99))  # 打印第一个元素99的索引
print(c_list.count(99))  # 99出现的次数

c_list.sort()
print(c_list)
c_list.sort(reverse=True)
print(c_list)
```

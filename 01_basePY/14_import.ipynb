# import引入模块
(就是类似C的include)

```python
# time属于标准库,不需要
import time

print(time.localtime())
```

```python
import time as t

print(t.localtime())
```

```python
from time import time, localtime

print(localtime())
print(time())
```

```python
from time import *

print(localtime())
print(time())
# 其他函数也可以直接用...
```

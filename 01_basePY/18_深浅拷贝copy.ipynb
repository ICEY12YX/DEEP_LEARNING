# 深浅拷贝copy

```python
import copy

a = [1, 2, 3]
b = a  # 复制房间号
print(id(a))
print(id(b))
print(id(a) == id(b))

b[0] = 8
print(a)
print(b)

a[1] = 8
print(a)
print(b)

c = copy.copy(a)
print(id(a))
print(id(c))
print(id(a) == id(c))
a[1] = 7
print(a)
print(c)

a = [1, 2, 3, [4, 5]]
d = copy.copy(a)
print(id(a))
print(id(d))
print(id(a) == id(d))
print(id(a[3]))
print(id(d[3]))
print(id(a[3]) == id(d[3]))
a[3][0] = 7
print(a)
print(d)

e = copy.deepcopy(a)
print(id(a))
print(id(e))
print(id(a) == id(e))
print(id(a[3]))
print(id(e[3]))
print(id(a[3]) == id(e[3]))
a[3][0] = 99
print(a)
print(e)
```

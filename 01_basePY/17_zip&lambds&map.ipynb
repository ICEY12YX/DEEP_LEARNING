# zip & lambdas & maps

```python
a = [1, 2, 3]
b = [4, 5, 6]
list(zip(a, b))  # 疑问: list是什么
print(zip(a, b))  # 直接输出,输出的是obj
print(list(zip(a, b)))  # 疑问: list(zip(a, b)) 和这句,为什么只有一个输出

for i, j in zip(a, b):
    print(i / 2, j * 2)

list(zip(a, a, b))
```

```python
def fun1(x, y):
    return x + y


print(fun1(1, 2))

# 简单函数可以用lambda
fun2 = lambda x, y: x + y
print(fun2(3, 4))
```

```python
def fun1(x, y):
    return x + y


# 疑问,为什么两行list只输出最后一行
list(map(fun1, [1, 2], [3, 4]))
list(map(fun1, [1, 2], [3]))

print(list(map(fun1, [1, 2], [3, 4])))
print(list(map(fun1, [1, 2], [3])))
```

```python
def functionA():  # 同样注意tab缩进, 缩进的是函数里面内容,没缩进的就不是
    print("functionA is running")
    a = 1 + 2
    print(a)


functionA()  # 调用函数, 执行函数里面的内容
```

```python
def functionB(x, y):  # 函数可以有参数, 也可以没有参数
    z = x + y
    print("the z is", z)  # 这个输出可以注意到,is后面自己带了空格
    print("the z is " + str(z))  # 注意两种print写法


functionB(1, 2)  # 调用函数, 并传入参数
functionB(x=4, y=7)  # 调用函数, 并传入参数
# functionB(a=1, b=2)  # 参数名不对, 会报错,参数名要和形参一致(这点不大能用C语言去思考)
```

```python
# def functionC(price, color="red", discount=0.8,text):
# 默认参数后面不能再有非默认参数, 也就是text参数不能放在后面, 会报错


def functionC(price, color="red", discount=0.8):
    # 函数可以有默认参数, 默认参数要放在最后面
    print("the price is", price)
    print("the color is", color)
    print("the discount is", discount)


functionC(100)  # 调用函数, 只传入一个参数, 其他参数使用默认值
functionC(100, color="blue")  # 调用函数, 默认参数也可以被覆盖
```

```python
APPLE = 88  # 全局变量


def functionD():
    a = 10
    b = APPLE
    print("the a is", a)
    print("the b is", b)

    # APPLE = 99
    # 疑问：为什么写这句会报错 local variable 'APPLE' referenced before assignment?
    # 意思是APPLE 被引用了 (在被赋值之前)
    return a + 10


# 函数可以有返回值, return后面可以跟一个值,也可以跟多个值,也可以不跟值
# python写函数的时候.可以不用考虑返回值类型,不像c输入输出都要先思考清楚

print(functionD())
# print(APPLE)
# # 全局变量可以在函数外部使用, 也可以在函数内部使用, 但是不能在函数内部修改全局变量的值
# # print(a)  # 函数内部的局部变量不能在函数外部使用,
```

```python
APPLE = 88


def functionD():
    global APPLE  # 第一步：举牌子声明
    a = 10
    b = APPLE  # 此时读取的是全局的 88
    APPLE = 99  # 此时修改的也是全局的 APPLE
    print("the a is", a)
    print("the b is", b)
    return a + 10


print(functionD())
```

```python
x = 1
y = 2
z = 3
if x < y:
    print("x is less than y")

if x < y < z:
    print("x is less than y, and y is less than z")

if x <= y:
    print("x is less than or equal to y")

if x == 1:  # 不能写x=1,一个等号是赋值语句
    print("x is equal to 1")
```

# RegEx正则表达式

```python
import re

pattern = "cat"
string = "cat"
print(pattern in string)
print(re.search(pattern, string))
# 需要先搞个pattern,用来匹配
# 找到就返回obj对象

# pattern也可以是表达式
# []表示或,中括号里的东西选一个
# [0-9]表示数字0到9中的一个
# search(pattern, string) 这里也可以直接把正则表达式填进去,不用pattern这个变量

# \d 表示数字0到9中的一个
# \D 表示非数字字符
# \s 表示空白字符 比如空格、制表符、换行符
# \S 表示非空白字符
# \w 表示字母数字下划线
# \W 表示非字母数字下划线
# \b 表示单词边界 (疑问 需要举例)
# \B 表示非单词边界 (疑问 需要举例)
# \\ 表示反斜杠本身
# . 表示任意字符（除了换行符）
# ^ 表示字符串的开头
# $ 表示字符串的结尾
# ? 表示前面的字符是可选的（出现0次或1次） 就是可以有或没有
# * 表示前面的字符可以出现任意次（包括0次）
# + 表示前面的字符至少出现一次
# {n,m} 表示前面的字符至少出现n次，至多出现m次

# flags=re.M 表示匹配多行

# group 表示捕获的分组，可以通过group(n)获取第n个分组的内容
# match好像和group是一起用的 (疑问)
# group可以有名字 ?P<xxx>

# findall 返回所有匹配的列表
# | 或者 or  表示或,例如 a|b 匹配 a 或 b
# re.sub(pattern, repl, string) 用repl替换string中所有匹配pattern的子串
# re.split() 按照匹配的子串将string分割成列表
# compile(pattern, flags=0) 编译正则表达式，返回一个正则表达式对象，可以提高匹配效率
# 疑问:complie什么用
```
